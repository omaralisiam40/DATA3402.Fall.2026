1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter.

```python
class counter:
    def __init__(self, maximum):
        self.maximum = maximum
        self.value = 0

    def increment(self):
        if self.value < self.maximum:
            self.value += 1
        else:
            print("Error: counter cannot exceed the maximum value.")

    def reset(self):
        self.value = 0
```

5. Implement a common base class for the classes implemented in 3 and 4 above which implements all common methods as not implemented functions (virtual). Re-implement your rectangle and circle classes to inherit from the base class and overload the functions accordingly.

```python
import math
from abc import ABC, abstractmethod

class shape(ABC):
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class circle(shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def area(self):
        return math.pi * (self.__radius ** 2)

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

class rectangle(shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width
```

6. Implement a triangle class analogous to the rectangle and circle in question 5.

```python
class triangle(shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)

        if (side_a <= 0 or side_b <= 0 or side_c <= 0 or
            side_a + side_b <= side_c or
            side_a + side_c <= side_b or
            side_b + side_c <= side_a):
            raise ValueError("The sides must form a valid triangle.")

        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.__side_a) *(s - self.__side_b) *(s - self.__side_c))

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the perimeter of the object.

```python
import math

def triangle_vertices(t):
    x, y = t.get_x(), t.get_y()
    a = t.get_side_a()
    b = t.get_side_b()
    c = t.get_side_c()

    third_x = (b*b + c*c - a*a) / (2*c)
    third_y = math.sqrt(max(0, b*b - third_x*third_x))

    return [(x, y), (x+c, y), (x+third_x, y+third_y)]


def points_on_edges(vertices, count):
    if not 1 <= count <= 16:
        raise ValueError("count must be between 1 and 16")

    edges = []
    for i in range(len(vertices)):
        start = vertices[i]
        end = vertices[(i+1) % len(vertices)]
        edges.append((start, end, math.dist(start, end)))

    total_length = sum(edge[2] for edge in edges)
    points = []

    for i in range(count):
        distance = i * total_length / count

        for start, end, length in edges:
            if distance <= length:
                fraction = distance / length
                px = start[0] + fraction * (end[0] - start[0])
                py = start[1] + fraction * (end[1] - start[1])
                points.append((px, py))
                break
            distance -= length

    return points


def base_perimeter_points(self, count=16):
    raise NotImplementedError("Each shape must provide perimeter points.")


def rectangle_perimeter_points(self, count=16):
    x, y = self.get_x(), self.get_y()
    length, width = self.get_length(), self.get_width()
    corners = [(x, y), (x+length, y),
               (x+length, y+width), (x, y+width)]
    return points_on_edges(corners, count)


def circle_perimeter_points(self, count=16):
    if not 1 <= count <= 16:
        raise ValueError("count must be between 1 and 16")

    return [ (self.get_x() + self.get_radius() * math.cos(2*math.pi*i/count), self.get_y() + self.get_radius() * math.sin(2*math.pi*i/count)) for i in range(count)]


def triangle_perimeter_points(self, count=16):
    return points_on_edges(triangle_vertices(self), count)


shape.perimeter_points = base_perimeter_points
rectangle.perimeter_points = rectangle_perimeter_points
circle.perimeter_points = circle_perimeter_points
triangle.perimeter_points = triangle_perimeter_points

```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
def point_in_polygon(px, py, vertices):
    signs = []

    for i in range(len(vertices)):
        x1, y1 = vertices[i]
        x2, y2 = vertices[(i+1) % len(vertices)]
        cross = (x2-x1)*(py-y1) - (y2-y1)*(px-x1)

        if abs(cross) > (1 * 10**-10):
            signs.append(cross > 0)

    return not signs or all(sign == signs[0] for sign in signs)


def base_contains(self, x, y):
    raise NotImplementedError("Each shape must test its own interior.")


def rectangle_contains(self, x, y):
    return (self.get_x() <= x <= self.get_x() + self.get_length()
            and self.get_y() <= y <= self.get_y() + self.get_width())


def circle_contains(self, x, y):
    distance = math.hypot(x-self.get_x(), y-self.get_y())
    return distance <= self.get_radius()


def triangle_contains(self, x, y):
    return point_in_polygon(x, y, triangle_vertices(self))


shape.contains = base_contains
rectangle.contains = rectangle_contains
circle.contains = circle_contains
triangle.contains = triangle_contains
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python
def polygon_vertices(obj):
    if isinstance(obj, rectangle):
        x, y = obj.get_x(), obj.get_y()
        length, width = obj.get_length(), obj.get_width()
        return [(x, y), (x+length, y),
                (x+length, y+width), (x, y+width)]

    if isinstance(obj, triangle):
        return triangle_vertices(obj)

    raise TypeError("Expected a rectangle or triangle")

def polygons_overlap(first, second):
    # Testing a perpendicular direction for every edge of both polygons.
    for polygon in (first, second):
        for i in range(len(polygon)):
            x1, y1 = polygon[i]
            x2, y2 = polygon[(i+1) % len(polygon)]
            axis_x, axis_y = -(y2-y1), x2-x1

            values1 = [x*axis_x + y*axis_y for x, y in first]
            values2 = [x*axis_x + y*axis_y for x, y in second]

            if (max(values1) < min(values2) - 1e-10 or
                max(values2) < min(values1) - 1e-10):
                return False

    return True

def circle_polygon_overlap(c, vertices):
    cx, cy = c.get_x(), c.get_y()

    if point_in_polygon(cx, cy, vertices):
        return True

    for i in range(len(vertices)):
        x1, y1 = vertices[i]
        x2, y2 = vertices[(i+1) % len(vertices)]
        dx, dy = x2-x1, y2-y1

        t = ((cx-x1)*dx + (cy-y1)*dy) / (dx*dx + dy*dy)
        t = max(0, min(1, t))

        closest_x = x1 + t*dx
        closest_y = y1 + t*dy

        if math.hypot(cx-closest_x, cy-closest_y) <= c.get_radius() + 1e-10:
            return True

    return False


def overlaps(self, other):
    if not isinstance(other, shape):
        raise TypeError("other must be a shape")

    if isinstance(self, circle) and isinstance(other, circle):
        distance = math.hypot(self.get_x()-other.get_x(),
                              self.get_y()-other.get_y())
        return distance <= self.get_radius() + other.get_radius()

    if isinstance(self, circle):
        return circle_polygon_overlap(self, polygon_vertices(other))

    if isinstance(other, circle):
        return circle_polygon_overlap(other, polygon_vertices(self))

    return polygons_overlap(polygon_vertices(self), polygon_vertices(other))


shape.overlaps = overlaps
```

```python
r = rectangle(5, 3, 0, 0)
c = circle(2, 4, 1)
t = triangle(3, 4, 5, 10, 0)

print(r.overlaps(c))
print(r.overlaps(t))
print(c.overlaps(t))
```

```python
%%writefile paint.py
import json
import math
from abc import ABC, abstractmethod
from PIL import Image, ImageDraw


class Canvas:
    def __init__(self, width=500, height=350, background="white"):
        self.image = Image.new("RGB", (width, height), background)
        self.pen = ImageDraw.Draw(self.image)

    def save(self, filename):
        self.image.save(filename)

    def show(self):
        display(self.image)


class shape(ABC):
    def __init__(self, x, y, color="black"):
        self.__x = x
        self.__y = y
        self.__color = color

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def get_color(self):
        return self.__color

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    @abstractmethod
    def paint(self, canvas):
        pass


class rectangle(shape):
    def __init__(self, length, width, x, y, color="blue"):
        super().__init__(x, y, color)
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def paint(self, canvas):
        x, y = self.get_x(), self.get_y()
        canvas.pen.rectangle(
            (x, y, x + self.__length, y + self.__width),
            fill=self.get_color(),
            outline="black"
        )


class circle(shape):
    def __init__(self, radius, x, y, color="red"):
        super().__init__(x, y, color)
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def area(self):
        return math.pi * self.__radius**2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def paint(self, canvas):
        x, y = self.get_x(), self.get_y()
        r = self.__radius
        canvas.pen.ellipse(
            (x - r, y - r, x + r, y + r),
            fill=self.get_color(),
            outline="black"
        )


class triangle(shape):
    def __init__(self, side_a, side_b, side_c, x, y, color="green"):
        if (side_a <= 0 or side_b <= 0 or side_c <= 0 or
            side_a + side_b <= side_c or
            side_a + side_c <= side_b or
            side_b + side_c <= side_a):
            raise ValueError("The sides must form a valid triangle.")

        super().__init__(x, y, color)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(
            s * (s - self.__side_a) *
            (s - self.__side_b) *
            (s - self.__side_c)
        )

    def paint(self, canvas):
        # side_c lies horizontally; the third corner is below it
        x, y = self.get_x(), self.get_y()
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        third_x = (b*b + c*c - a*a) / (2*c)
        third_y = math.sqrt(max(0, b*b - third_x*third_x))

        canvas.pen.polygon(
            [(x, y), (x + c, y), (x + third_x, y + third_y)],
            fill=self.get_color(),
            outline="black"
        )


class CompoundShape:
    def __init__(self):
        self.shapes = []

    def add(self, obj):
        self.shapes.append(obj)

    def remove(self, obj):
        self.shapes.remove(obj)

    def paint(self, canvas):
        for obj in self.shapes:
            obj.paint(canvas)


class RasterDrawing(CompoundShape):
    def __init__(self, width=500, height=350):
        super().__init__()
        self.width = width
        self.height = height

    def paint(self, filename=None):
        canvas = Canvas(self.width, self.height)
        super().paint(canvas)

        if filename is not None:
            canvas.save(filename)

        canvas.show()
        return canvas

    def save(self, filename):
        data = {
            "width": self.width,
            "height": self.height,
            "shapes": []
        }

        for obj in self.shapes:
            if isinstance(obj, rectangle):
                item = {
                    "type": "rectangle",
                    "length": obj.get_length(),
                    "width": obj.get_width()
                }
            elif isinstance(obj, circle):
                item = {
                    "type": "circle",
                    "radius": obj.get_radius()
                }
            elif isinstance(obj, triangle):
                item = {
                    "type": "triangle",
                    "side_a": obj.get_side_a(),
                    "side_b": obj.get_side_b(),
                    "side_c": obj.get_side_c()
                }
            else:
                raise TypeError("Unknown shape type")

            item["x"] = obj.get_x()
            item["y"] = obj.get_y()
            item["color"] = obj.get_color()
            data["shapes"].append(item)

        with open(filename, "w") as file:
            json.dump(data, file, indent=2)


def load_drawing(filename):
    with open(filename, "r") as file:
        data = json.load(file)

    drawing = RasterDrawing(data["width"], data["height"])

    for item in data["shapes"]:
        if item["type"] == "rectangle":
            obj = rectangle(
                item["length"], item["width"],
                item["x"], item["y"], item["color"]
            )
        elif item["type"] == "circle":
            obj = circle(
                item["radius"], item["x"], item["y"], item["color"]
            )
        elif item["type"] == "triangle":
            obj = triangle(
                item["side_a"], item["side_b"], item["side_c"],
                item["x"], item["y"], item["color"]
            )
        else:
            raise ValueError("Unknown shape type in file")

        drawing.add(obj)

    return drawing
```

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b

    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"

    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()


def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```

```python
# Test
print(repr(foo(1,"hello")))
```

```python
# Create an object and save it
ff=foo(1,"hello")
ff.save("Test.foo")
```

```python
# Check contents of the saved file
!cat Test.foo
```

```python
# Load the object
ff_reloaded=foo_loader("Test.foo")
ff_reloaded
```
