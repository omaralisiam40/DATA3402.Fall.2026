*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def get_winner(board):
    n = len(board)

    # Checking every row
    for row in board:
        if row[0] != 0 and all(value == row[0] for value in row):
            return row[0]

    # Checking every column
    for column in range(n):
        first_value = board[0][column]
        winner = True

        for row in range(n):
            if board[row][column] != first_value:
                winner = False

        if first_value != 0 and winner:
            return first_value

    # Checking the diagonal from top-left to bottom-right
    first_value = board[0][0]
    winner = True

    for i in range(n):
        if board[i][i] != first_value:
            winner = False

    if first_value != 0 and winner:
        return first_value

    # Checking the diagonal from top-right to bottom-left
    first_value = board[0][n - 1]
    winner = True

    for i in range(n):
        if board[i][n - 1 - i] != first_value:
            winner = False

    if first_value != 0 and winner:
        return first_value

    # Checking whether the game is incomplete
    for row in board:
        if 0 in row:
            return -1

    # No winner and no empty spaces means a draw
    return 0
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
# Write your solution here
def get_location(board, player):
    while True:
        location = input(f"Player {player}, enter a location: ")

        if len(location) != 2 or not location[0].isalpha() or not location[1].isdigit():
            print("Invalid location. Please enter a letter and a number.")
            continue

        row = int(location[1]) - 1
        col = ord(location[0].upper()) - 65

        if row < 0 or row >= len(board) or col < 0 or col >= len(board[0]):
            print("Invalid location. Please enter a valid location.")
            continue

        if not make_move_location(board, player, location):
            print("Invalid location. Please enter a valid location.")
            continue

        return location
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
def computer_move(board):
    n = len(board)

    # First, try to win
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = 2

                if get_winner(board) == 2:
                    return True

                board[row][col] = 0

    # If it cannot win, block player 1
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = 1

                if get_winner(board) == 1:
                    board[row][col] = 2
                    return True

                board[row][col] = 0

    # Take the center if it is available
    center = n // 2

    if board[center][center] == 0:
        board[center][center] = 2
        return True

    # Otherwise, take the first available location
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = 2
                return True

    return False


def play_computer_game(n):
    board = make_board(n)

    while True:
        # Human player's turn
        draw_board(board)
        get_location(board, 1)

        result = get_winner(board)

        if result == 1:
            draw_board(board)
            print("Player 1 (X) wins!")
            return 1

        if result == 0:
            draw_board(board)
            print("The game is a draw!")
            return 0

        # Computer's turn
        computer_move(board)
        print("The computer made its move.")

        result = get_winner(board)

        if result == 2:
            draw_board(board)
            print("Computer (O) wins!")
            return 2

        if result == 0:
            draw_board(board)
            print("The game is a draw!")
            return 0

````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def search_board(board, depth, maximizing):
    result = get_winner(board)

    # Terminal results
    if result == 2:
        return 10 + depth
    elif result == 1:
        return -10 - depth
    elif result == 0:
        return 0

    # Stop searching at the maximum depth
    if depth == 0:
        return 0

    n = len(board)

    # Computer's turn: maximize the score
    if maximizing:
        best_score = -1000

        for row in range(n):
            for col in range(n):
                if board[row][col] == 0:
                    board[row][col] = 2
                    score = search_board(board, depth - 1, False)
                    board[row][col] = 0

                    if score > best_score:
                        best_score = score

        return best_score

    # Human's turn: minimize the score
    else:
        best_score = 1000

        for row in range(n):
            for col in range(n):
                if board[row][col] == 0:
                    board[row][col] = 1
                    score = search_board(board, depth - 1, True)
                    board[row][col] = 0

                    if score < best_score:
                        best_score = score

        return best_score

def smart_computer_move(board, max_depth):
    n = len(board)
    best_score = -1000
    best_row = -1
    best_col = -1

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                # Temporarily try this move
                board[row][col] = 2

                score = search_board(
                    board,
                    max_depth - 1,
                    False
                )

                # Undo the temporary move
                board[row][col] = 0

                if score > best_score:
                    best_score = score
                    best_row = row
                    best_col = col

    if best_row != -1:
        board[best_row][best_col] = 2
        return True

    return False



def play_smart_computer_game(n, max_depth):
    board = make_board(n)

    while True:
        draw_board(board)

        # Human is player 1
        get_location(board, 1)

        result = get_winner(board)

        if result == 1:
            draw_board(board)
            print("Player 1 (X) wins!")
            return 1

        if result == 0:
            draw_board(board)
            print("The game is a draw!")
            return 0

        # Computer is player 2
        smart_computer_move(board, max_depth)
        print("The computer made its move.")

        result = get_winner(board)

        if result == 2:
            draw_board(board)
            print("Computer (O) wins!")
            return 2

        if result == 0:
            draw_board(board)
            print("The game is a draw!")
            return 0
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
import random
def simple_computer_move(board, player):
    n = len(board)

    if player == 1:
        opponent = 2
    else:
        opponent = 1

    # Try to win
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = player

                if get_winner(board) == player:
                    return True

                board[row][col] = 0

    # Block the opponent
    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = opponent

                if get_winner(board) == opponent:
                    board[row][col] = player
                    return True

                board[row][col] = 0

    # Take the center
    center = n // 2

    if board[center][center] == 0:
        board[center][center] = player
        return True

    # Otherwise choose a random empty location
    empty_locations = []

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                empty_locations.append((row, col))

    if len(empty_locations) > 0:
        row, col = random.choice(empty_locations)
        board[row][col] = player
        return True

    return False


def compare_computers(number_of_games, max_depth):
    simple_wins = 0
    smart_wins = 0
    draws = 0

    for game in range(number_of_games):
        board = make_board(3)

        # Alternate which computer starts
        if game % 2 == 0:
            current_player = 1
        else:
            current_player = 2

        while get_winner(board) == -1:
            if current_player == 1:
                simple_computer_move(board, 1)
            else:
                smart_computer_move(board, max_depth)

            result = get_winner(board)

            if result != -1:
                break

            if current_player == 1:
                current_player = 2
            else:
                current_player = 1

        if result == 1:
            simple_wins += 1
        elif result == 2:
            smart_wins += 1
        else:
            draws += 1

        print("Game", game + 1, "result:", result)

    simple_win_rate = simple_wins / number_of_games * 100
    smart_win_rate = smart_wins / number_of_games * 100
    draw_rate = draws / number_of_games * 100

    print()
    print("Simple computer wins:", simple_wins)
    print("Smart computer wins:", smart_wins)
    print("Draws:", draws)
    print("Simple computer win rate:", simple_win_rate, "%")
    print("Smart computer win rate:", smart_win_rate, "%")
    print("Draw rate:", draw_rate, "%")
````

````python
# Test your solution here
compare_computers(10, 9)
````
